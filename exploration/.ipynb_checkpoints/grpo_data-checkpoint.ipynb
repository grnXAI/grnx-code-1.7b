```python
import pandas as pd
import torch
from torch import nn
import os
from tqdm import tqdm
```

```python
shards = {}
path = "/Users/gromeronaranjo/Desktop/grnx-code-1.7b/data/raw/grpo/codecontests-plus-1x"

files = [
    file
    for file in os.listdir(path)
    if file.endswith(".parquet")
]

for idx, file in tqdm(
    enumerate(files),
    total=len(files)
):
    shards[f"shard_{idx}"] = os.path.join(path, file)

shard = pd.read_parquet(shards["shard_1"])

shard.head()
```
